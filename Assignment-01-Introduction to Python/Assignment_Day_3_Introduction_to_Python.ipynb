## Soal 2: Membuat Input
Buatlah dua kode agar pengguna bisa melakukan input untuk data berikut:
*   `Nama produk`
*   `Jumlah produk terjual`
*   `Harga produk`





````python
# Cara 1 menggunakan Fungsi
def ambil_input_produk():
  nama = input('Masukan Nama Produk: ')
  jumlah = int(input('Masukan Jumlah Terjual: '))
  harga = int(input('Masukan Harga Produk (Rp): '))
  return nama, jumlah, harga

# Memanggil fungsi untuk mengambil input
nama_produk, jumlah_produk_terjual, harga_produk = ambil_input_produk()

print(f'Nama Produk: {nama_produk}, Jumlah Terjual: {jumlah_produk_terjual}, Harga Produk: {harga_produk}')
````

````python
# Cara 2 Manual Input
nama_produk = input('Masukan nama produk: ')
jumlah_produk_terjual = int(input('Masukan jumlah terjual'))
harga_produk = int(input('Masukan harga produk (Rp): '))

print('Nama Produk :', nama_produk)
print('Jumlah Produk Terjual :', jumlah_produk_terjual)
print('Harga Produk :', harga_produk)
````

## Soal 3: Logika If/Else untuk Rekomendasi Produk
Langkah:
1. Hitung **total revenue** = `jumlah_produk_terjual x harga_produk`.
2. Gunakan `if / elif / else` untuk menentukan kategori penjualan:
   - `total_revenue <= 2.000.000` → **"Rendah"**
   - `2.000.000 < total_revenue <= 8.000.000` → **"Sedang"**
   - `total_revenue > 8.000.000` → **"Tinggi"**

````python
# Menghitung total revenue
total_revenue = jumlah_produk_terjual * harga_produk

# Menentukan kategori penjualan
if total_revenue <= 2000000:
  kategori_penjualan = 'rendah'
elif total_revenue <= 8000000:
  kategori_penjualan = 'sedang'
else:
  kategori_penjualan = 'tinggi'

print(f'total revenue: {total_revenue}, kategori penjualan: {kategori_penjualan}')
````

## Soal 4: Output Akhir Program
Tampilkan rekomendasi hasil analisis penjualan yang meliputi:
   - `Nama Produk`
   - `Jumlah Produk`
   - `Harga Produk`
   - `Total Revenue`
   - `Kategori Penjualan (Rendah, sedang, tinggi)`
   
**Goals**: Menampilkan rekomendasi produk

````python
if kategori_penjualan == 'rendah':
  rekomendasi = 'Penjualan rendah, disarankan memperbaiki marketing/ mengganti produk'
elif kategori_penjualan == 'sedang':
  rekomendasi = 'Penjualan stabil, pertahankan marketing kalau bisa tingkatkan'
else:
  rekomendasi = 'penjualan sangat baik, pertahankan'

print('Hasil Analisis Penjualan')
print(f'nama produk: {nama_produk}')
print(f'jumlah produk: {jumlah_produk_terjual}')
print(f'harga produk: {harga_produk}')
print(f'total revenue: {total_revenue}')
print(f'kategori penjualan: {kategori_penjualan}')
print(f'rekomendasi: {rekomendasi}')
````

## Soal 5: Penjelasan Logika Program

**Bagaimana total revenue dihitung?**

Total revenue dapat dilihat dengan cara mengkalikan `jumlah_produk_terjual` dengan `harga_produk`, semakin banyak barang yang terjual/ semakin tinggi barang yang dijual maka semakin tinggi revenue yang didapat. Didalam kode dapat ditulis seperi:

```total revenue= jumlah_produk_terjual * harga_produk```


**Bagaimana if/else digunakan menentukan kategori penjualan?**

`if/else` dalam pengecekan `total_revenue` akan terus-terusan sampai dia menemukan kondisi yang sesuai dari atas sampai bawah.
1.   `if total_revenue <= 2.000.000` , jika kondisi sesuai disini maka program akan memberi label **"rendah"** dan berhenti mengecek kondisi yang lain.
2.   Jika tidak sesuai kondisi kode pertama, program akan lanjut mengecek `elif total_revenue <= 8.000.000` dan jika kondisi sesuai kondisi ini makan program akan memberikan label **"sedang"**
3.   Jika 2 kondisi diatas salah berarti total revenue > 8.000.000, maka program akan masuk ke `else` dan memberi label **"tinggi"**

Jadi kita dapat mengetahui kondisi penjualan suatu produk baik sedang kuat/ lemah.
