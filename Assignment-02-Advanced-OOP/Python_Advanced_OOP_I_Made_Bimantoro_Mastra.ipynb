# Soal 1: Membuat Class dan Object
Buatlah class dan object dengan ketentuan sebagai berikut:

1.   Sebuah class bernama Mahasiswa
2.   Tambahkan tiga atribut di dalam class tersebut, yaitu: **nama**, **nim**, dan **jurusan**.
1.   Menampilkan informasi mahasiswa (nama, nim, dan jurusan).



```python
class Mahasiswa:
  def __init__(self, nama, nim, jurusan):
    self.nama = nama
    self.nim = nim
    self.jurusan = jurusan

  def info_mahasiswa(self):
    print(f"Nama: {self.nama}")
    print(f"NIM: {self.nim}")
    print(f"Jurusan: {self.jurusan}")
    return f'Mahasiswa bernama {self.nama} dengan NIM {self.nim} dari jurusan {self.jurusan}'

 # Testing
mahasiswa1 = Mahasiswa('Rover', '2440038433', 'Machine Learning')
print(mahasiswa1.info_mahasiswa())
```

dengan consturctor `__init__` setiap objek `Mahasiswa` mempunyai data sendiri tanpa perlu diset manual satu per satu. Untuk menampilkan datanya cukup menggunakan code `info_mahasiswa` karna python membaca object tersebut sebagai `self`

# Soal 2: Mengimplementasikan Encapsulation



1.   Buatlah sebuah class bernama **Tabungan**.
1.   Tambahkan sebuah atribut **saldo** yang bersifat **private** di dalam class tersebut.
2.   Buatlah method **deposit** yang berfungsi menambah saldo sesuai jumlah yang diberikan.
2.   Buatlah method **tarik** untuk **mengurangi saldo**.
1.   Berikan pengecekan di dalam metode tarik sehingga:
*   Jika saldo **tidak mencukupi**, tampilkan pesan: "**Saldo tidak mencukupi**."
*   Jika saldo mencukupi, proses pengurangan saldo dapat dilakukan.
6.   **Ujilah** class yang kamu buat dengan membuat minimal satu **object Tabungan**, kemudian **panggil** metode deposit dan tarik untuk melihat apakah fungsinya berjalan sesuai ketentuan.



```python
class Tabungan:
  def __init__(self, saldo):
    self.__saldo = saldo  #private

  def deposit(self, jumlah):
    self.__saldo += jumlah

  def tarik(self, jumlah):
    if self.__saldo < jumlah:
      print("Saldo tidak mencukupi.")
    else:
      self.__saldo -= jumlah
      print(f"Penarikan dengan {jumlah} berhasil, sisa saldo: {self.__saldo}")

  def cek_saldo(self):
    return self.__saldo
```

```python
# Test
rekening = Tabungan(100000)
rekening.cek_saldo()
```

```python
rekening.deposit(50000)
rekening.cek_saldo()
```

```python
rekening.tarik(20000)
rekening.cek_saldo()
```

```python
rekening.tarik(200000)
rekening.cek_saldo()
```

`__saldo private`, kode di luar class tidak bisa sembarangan mengubah saldo tanpa lewat deposit/tarik.

## Soal 3: Mengimplementasikan Inheritance


1.   Buatlah sebuah class bernama **Pegawai**.
1.   Tambahkan dua atribut di dalam class Pegawai, yaitu: nama dan gaji.
1.   Buatlah sebuah subclass bernama **Manager** yang **mewarisi** class Pegawai.
1.   Tambahkan sebuah atribut baru pada class Manager, yaitu tunjangan.
2.   Buatlah metode **info_manager** pada class Manager yang berfungsi **menampilkan informasi nama, gaji, dan tunjangan** dari instance Manager.
2.   Pastikan subclass Manager dapat mengakses atribut yang berasal dari class induk (Pegawai).
2.   Ujilah class Manager dengan membuat minimal satu **instance** Manager, kemudian **panggil** metode info_manager.




```python
class pegawai:
  def __init__(self, nama, gaji):
    self.nama = nama
    self.gaji = gaji

class manager(pegawai):
  def __init__(self, nama, gaji, tunjangan):
     super().__init__(nama, gaji)
     self.tunjangan = tunjangan

  def info_manager(self):
    print(f"Nama: {self.nama}")
    print(f"Gaji: {self.gaji}")
    print(f"Tunjangan: {self.tunjangan}")
    return f'Manager bernama {self.nama} dengan gaji {self.gaji} dan tunjangan {self.tunjangan}'

# Test
manager1 = manager('Rover', 10000000, 5000000)
print(manager1.info_manager())
```

Metode **inheritance** menggunakan code `super()` agar tidak perlu menuliskan sifat induknya lagi di anaknya.

# Soal 4: Polymorphism dan Method Overriding


1.   Buatlah sebuah class bernama **Kendaraan**.
1.   Tambahkan sebuah method bernama **bahan_bakar** di dalam class Kendaraan yang hanya menampilkan pesan:  **"Jenis bahan bakar tidak diketahui."**
1.   Buatlah subclass bernama **Mobil** yang **mewarisi** class Kendaraan.
1.   Buatlah subclass bernama **Motor** yang juga **mewarisi** class Kendaraan
2.   **Override** method **bahan_bakar** pada class Mobil sehingga menampilkan jenis bahan bakar yang digunakan (misalnya **"Bensin"**).
2.   **Override** method **bahan_bakar** pada class Motor untuk menampilkan jenis bahan bakarnya (misalnya "**Solar**", atau sesuai arahan soal).
2.   **Ujilah** ketiga class tersebut dengan membuat instance dari **Kendaraan, Mobil, dan Motor**, kemudian **panggil** method bahan_bakar pada masing-masing object untuk melihat perbedaannya.

```python
class kendaraan:
  def bahan_bakar(self):
    print("Jenis bahan bakar tidak diketahui.")

class mobil(kendaraan):
  def bahan_bakar(self):
    print("Bensin")

class motor(kendaraan):
  def bahan_bakar(self):
    print("Solar")

# Test
kendaraan_list = [kendaraan(), mobil(), motor()]
for k in kendaraan_list:
    k.bahan_bakar()
```

Loop menjalan method bahan_bakar() pada setiap objek yang berada di kendaraan_list() satu persatu. for k in `kendaraan_list`: artinya "ambil setiap item dari `kendaraan_list` satu per satu, dan sebut item itu k selama proses berlangsung".

# Soal 5: Abstraction dan Interfaces


1.   Buatlah sebuah **class abstrak** bernama **Hewan**.
1.   Tambahkan sebuah **method abstrak** bernama **suara** di dalam class Hewan.
  **Notes**: Method ini harus dideklarasikan sebagai abstrak, tanpa implementasi.
1.   Buatlah dua buah subclass bernama **Kucing** dan **Anjing** yang mewarisi class Hewan.
2.   Implementasikan method **suara** pada class Kucing dan Anjing sehingga menghasilkan suara khas hewan tersebut (contoh: "**Meow**" untuk kucing, **“Bark”** untuk anjing).
2.   Ujilah kedua subclass dengan membuat instance dari Kucing dan Anjing, lalu **panggil** method suara untuk memastikan implementasi berjalan sesuai ketentuan.


```python
from abc import ABC, abstractmethod

class hewan(ABC):
    @abstractmethod
    def suara(self):
        pass  #karna tidak ada suara yang sama antara jenis hewan jadi harus di override subclass

class kucing(hewan):
    def suara(self):
        print('meow')

class anjing(hewan):
    def suara(self):
        print('bark')

# Test
kucing = kucing()
kucing.suara()

anjing = anjing()
anjing.suara()
```

Hewan sebagai `class` abstrak dan suara sebagai `method abstrak`. Subclass kucing dan anjing harus mengoverride method menjadi punya mereka masing tiap subclass dikarenakan semua jenis hewan pasti mempunyai method suara tetapi isinya berbeda beda
